```python
import pandas as pd

txn = pd.read_csv(r"D:\PROJECTS\customer-churn-prediction\transactions_v2.csv\data\churn_comp_refresh\transactions_v2.csv",
                  parse_dates=['transaction_date','membership_expire_date'])
print(txn.shape)
txn.head()
```

```python
txn_sorted = txn.sort_values('transaction_date')

txn_agg = txn_sorted.groupby('msno').agg(
    last_price        = ('plan_list_price', 'last'),
    last_paid         = ('actual_amount_paid', 'last'),
    last_auto_renew   = ('is_auto_renew', 'last'),
    last_cancel       = ('is_cancel', 'last'),
    total_cancels     = ('is_cancel', 'sum'),
    num_transactions  = ('transaction_date', 'count'),
    last_expire       = ('membership_expire_date', 'last'),
).reset_index()

print(txn_agg.shape)
txn_agg.head()
```

```python
txn_agg.to_csv('txn_agg.csv', index=False)
print("Saved txn_agg.csv")
```
